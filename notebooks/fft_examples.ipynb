```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
import scienceplots
# plt.style.use(['science','std-colors'])
plt.style.use(['science','ieee'])
```

```python
def plot_triangle(center: float, width: float, height: float, color: str = None, linestyle=None):	
    plt.plot([center - width/2, center, center + width/2], [0, height, 0], color=color, linestyle=linestyle)
```

# Aliasing

```python
plt.figure(figsize=(5,2))
plot_triangle(0, 20, 1, color="k")
plt.ylabel("$|X_c(f)|$")
plt.xlabel("f (Hz)")
plt.xlim([-60, 60])
plt.ylim([0, 1.19])
plt.savefig(f"figures_dsp/example_aliasing_a.png", dpi=300)
```

```python
plt.figure(figsize=(5,2))
plot_triangle(0, 20, 1, color="black")
plot_triangle(25, 20, 1, color="gray", linestyle="--")
plot_triangle(-25, 20, 1, color="gray", linestyle="--")
plot_triangle(50, 20, 1, color="gray", linestyle="--")
plot_triangle(-50, 20, 1, color="gray", linestyle="--")
plt.ylabel("$|X_s(f)|$")
plt.xlabel("f (Hz)")
plt.xlim([-60, 60])
plt.ylim([0, 1.19])
plt.legend(["Original", "Replicas"], loc="upper right", ncols=2)
plt.savefig(f"figures_dsp/example_aliasing_b.png", dpi=300)
```

```python
plt.figure(figsize=(5,2))
plot_triangle(0, 20, 1, color="black")
plot_triangle(15, 20, 1, color="gray", linestyle="--")
plot_triangle(-15, 20, 1, color="gray", linestyle="--")
plot_triangle(30, 20, 1, color="gray", linestyle="--")
plot_triangle(-30, 20, 1, color="gray", linestyle="--")
plot_triangle(45, 20, 1, color="gray", linestyle="--")
plot_triangle(-45, 20, 1, color="gray", linestyle="--")
plt.ylabel("$|X_s(f)|$")
plt.xlabel("f (Hz)")
plt.xlim([-50, 50])
plt.ylim([0, 1.19])
plt.legend(["Original", "Replicas"], loc="upper right", ncols=2)
plt.savefig(f"figures_dsp/example_aliasing_c.png", dpi=300)
```

# Decimation and interpolation

```python
plt.figure(figsize=(5,2))

plot_triangle(0, 1, 1, color="k", linestyle="-")

plot_triangle(0, 2, 0.5, color="gray", linestyle="--")
plot_triangle(2, 2, 0.5, color="gray", linestyle="--")
plot_triangle(-2, 2, 0.5, color="gray", linestyle="--")

plt.ylabel("$|X(e^{j\omega})|$")
plt.xlabel("$\omega$ ($\pi$ rad)")
plt.xlim([-3, 3])
plt.ylim([0, 1.2])

# plt.xticks([-60, -40, -20, 0, 20, 40, 60]);
plt.yticks([0, 0.5, 1]);
plt.legend(["Original", "Downsampled"], loc="upper right", ncols=2)

plot_triangle(2, 1, 1, color="k", linestyle="-")
plot_triangle(-2, 1, 1, color="k", linestyle="-")

plt.savefig(f"figures_dsp/example_decimation_a.png", dpi=300)
```

```python
plt.figure(figsize=(5,2))

plot_triangle(0, 1, 1, color="k", linestyle="-")

plot_triangle(0, 0.5, 1, color="gray", linestyle="--")
for k in range(1,10):
    plot_triangle(k, 0.5, 1, color="gray", linestyle="--")
    plot_triangle(-k, 0.5, 1, color="gray", linestyle="--")

plt.ylabel("$|X(e^{j\omega})|$")
plt.xlabel("$\omega$ ($\pi$ rad)")

# plt.xticks([-k*10 for k in range(-7, 6)]);
plt.yticks([0, 0.5, 1]);
plt.legend(["Original", "Upsampled"], loc="upper right", ncols=2)
plt.xlim([-3, 3])
plt.ylim([0, 1.2])
plot_triangle(2, 1, 1, color="k", linestyle="-")
plot_triangle(-2, 1, 1, color="k", linestyle="-")

plt.savefig(f"figures_dsp/example_decimation_b.png", dpi=300)
```

```python
w = np.arange(-np.pi, np.pi, 0.01)
N = 8
W = np.exp(-1j*w*(N-1)/2) * np.sin(w * N/2) / np.sin(w/2)

plt.figure(figsize=(5,2))
plt.plot(w/np.pi, np.abs(W))
plt.grid(False)
plt.yticks([0, N], [0, "N"]);
plt.xlabel("$\omega$ ($\pi$ rad)")
plt.ylabel("$|W(e^{j\omega})|$")

plt.savefig(f"figures_dsp/example_window_spectrum.png", dpi=300)
```

```python
w = 0.5
N = 64
n = np.arange(N)
x = np.sin(w*n)
X = np.fft.fft(x)
f = np.linspace(-np.pi, np.pi, N)
plt.figure(figsize=(5,2))
plt.plot(f, np.abs(X), marker="x")
plt.grid(False)
```
