```python
import sys
import os
sys.path.append(".")
sys.path.append("..")
sys.path.append("...")

import numpy as np
import pandas as pd
import json
import pyarrow
from pathlib import Path
```

```python
path = Path("../../data/statistical_features/sequence_3/")
output_path = Path("../../data/statistical_features/")
files = os.listdir(path)
```

```python
# N = 10
```

```python
len(files)
```

```python
full_df = []
for file in files:
    df = pd.read_json(path / file, lines=True)
    full_df.append(df)

full_df = pd.concat(full_df)
full_df = pd.json_normalize(full_df.to_dict(orient="records"))
```

```python
full_df.to_csv(output_path / "sequence_3.csv", index=False)
```
