```python
from pathlib import Path
import sys
sys.path.append("..")
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
from scipy.signal import welch, csd, coherence
```

```python
path = Path("../data/time_examples/2022_11_19_08_34_RAW.txt")
```

```python
header = [
    "Sf[N]","c4t2_z:+Z[g]","c4t2_y:+Y[g]","c4t2_x:+X[g]","c3t2_z:+Z[g]","c3t2_y:+Y[g]","c3t2_x:+X[g]","c2t2_z:+Z[g]","c2t2_y:+Y[g]","c2t2_x:+X[g]","c1t2_z:+Z[g]","c1t2_y:+Y[g]","c1t2_x:+X[g]","Input_T2_Acc:+X[g]","Input_T2_F:+X[N]","S4_T4[N]","S3_T2[N]","S2_T2[N]","S1_T2[N]","Ref_S[N]"
]
```

```python
data = pd.read_table(path, skiprows=9, sep="    ", header=None)
data.columns = header
```

```python
a0 = data["Input_T2_Acc:+X[g]"]
f0 = data["Input_T2_F:+X[N]"]
x0 = data["c4t2_z:+Z[g]"]

N = len(a0)
fs = 256
dt = 1 / fs
```

```python
nperseg = 2048
inp = f0

xx_f, xx = welch(inp, fs, nperseg=nperseg)
yy_f, yy = welch(x0, fs, nperseg=nperseg)
xy_f, xy = csd(x0, inp, fs=fs, nperseg=nperseg)

h1 = xy / xx
h2 = yy / xy
coh_f, coh_x = coherence(inp, x0, fs, nperseg=nperseg)
```

```python
px.line(x=yy_f, y=np.abs(yy))
```

```python
px.line(x=xy_f, y=np.abs(h2))
```

```python
px.line(x=coh_f, y=np.abs(coh_x))
```
