```python
PARAMS = {
    "level": 4,
    "family": "db16",
    "d": 2,
}
```

```python
level = PARAMS["level"]
family = PARAMS["family"]
d = PARAMS["d"]
```

```python
PATH_TO_VALID_DATASET = r"../raw/sequence_1/"
PATH_TO_TRAIN_DATASET = r"../raw/sequence_2/"
PATH_TO_TEST_DATASET = r"../raw/sequence_3/"

PATH_MODELS = r"../models/metric_learning/models_repo"
```

```python
POSSIBLE_CABLES = [1,2,3,4]
```

```python
transformer = WPTFeatureExtractor(
    level = level,
    family = family,
    reduction = "energy",
    normalize = True,
)
```

```python
features_per_cable_train = load_transform(PATH_TO_TRAIN_DATASET, transformer)
```

```python
for cable in range(1,5):
    print(cable)
    cols = [c for c in features_per_cable_train[f"c{cable}"].columns if isinstance(c, int)]
    x_wpt = features_per_cable_train[f"c{cable}"][cols].to_numpy()
    reducer = MLPFeatureReducer(
        N=x_wpt.shape[1],
        d=2,
        hidden=32
    )
    reducer_scikit = FeatureReducerScikit(
        reducer_model=reducer,
        loss_fn = TripletMarginLoss(margin=1, p=2)
    )
    reducer_scikit.fit(x_wpt, num_epochs=50)

    output = {
        "feature_extraction": transformer,
        "feature_reduction": reducer,
        "params": PARAMS
    }
    filename = f"c{cable}_l{level}_f{family}_d{d}.pkl"

    with open(f"{PATH_MODELS}/{filename}", "wb") as h:
        pickle.dump(output, h, protocol=pickle.HIGHEST_PROTOCOL)
    
    del output
```
